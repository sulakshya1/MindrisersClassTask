### Data Type
A data type defines the kind of value a variable holds and the operations that can be performed on that value.<br>
For example, Python treats 10, 10.5, and "10" differently, even though they look similar.

#### Common Python data types
|Datatype|Description|Example|
|--------|-----------|-------|
|int |Integer numbers|	10|
|float	|Decimal numbers	|3.14|
|str|Text|"Python"|
|bool	|True or false|True|
|list	|Ordered, changeable collection	|[1, 2, 3]|
|tuple	|Ordered, unchangeable collection	|(1, 2, 3)|
|dict	|Key-value collection	|{"name": "Ram"}|
|set	|Collection of unique values	|{1, 2, 3}|

```python
a = 10
price = 99.50
complex_num = 2 + 3j
is_student = True
name = "Ram"
fruits = ["apple", "banana", "mango"]
coordinates = (10, 20, 30)
numbers = range(1, 6)
set_numbers = {1, 2, 2, 3, 4}
```

```python
student = {
    "name": "Ram",
    "age": 20,
    "course": "Python"
}
```

```python
print(type(a))
print(type(price))
print(type(complex_num))
print(type(is_student))
print(type(name))
print(type(fruits))
print(type(coordinates))
print(type(numbers))
print(type(student))
print(type(set_numbers))
```

### Type Casting in Python
Type casting means converting a value from one data type to another.<br>
For example, you might convert text containing digits into an integer so that you can perform calculations.<br>
Python supports built-in conversion functions such as int( ), float( ), str( ), bool( ), list( ), tuple( ), set( ), and dict( ).

#### Types of type conversion
Implicit type conversion: Python automatically converts a value's type in certain operations.<br>

Explicit type conversion: The programmer uses a function to request conversion.

```python
a = 10
b = 2.5

result = a + b
print(result)
print(type(result))
```

```python
age = "20"
age = int(age)
print(age + 5)
```

#### Common type-casting functions
|Function	|Conversion	|Example	|Result|
|-----------|-----------|-----------|------|
|int( )	|To integer	|int("25")	|25|
|float( )|	To float	|float("3.5")	|3.5|
|str( )	|To string	|str(100)	|"100"|
|bool( )|To Boolean	|bool(1)	|True|
|list( )	|To list	|list((1, 2))	|[1, 2]|
|tuple( )	|To tuple	|tuple([1, 2])	|(1, 2)|
|set( )	|To set	|set([1, 1, 2])	|{1, 2}|

Converting a string to an integer

```python
number = "100"

number = int(number)
print(number + 50)
```

Converting a number to a string

```python
age = 20

message = "My age is " + str(age)
print(message)
```

Converting a float to an integer

```python
print(int(9.8))
print(int(-9.8))
print(round(9.8))
```

Converting user input

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(a + b)
```

Boolean type casting

```python
print(bool(1))
print(bool(0))
print(bool(""))
print(bool("Python"))
```
