### Functions 
A function is a reusable block of code that performs a specific task.<br>
Instead of writing the same code repeatedly, you write it once inside a function and call it whenever needed.

```python
def greet():
    print("Hello, welcome to Python!")
```

```python
greet()
greet()
```

#### Parameters and arguments
These two terms are related but have different meanings.
- Parameter: A variable listed in the function definition.
- Argument: The actual value supplied when calling the function.

```python
def greet(name):       # name is a parameter
    print("Hello", name)
```

```python

greet("Ram")           # "Ram" is an argument
greet("Sita")          # "Sita" is an argument
```

#### Functions with multiple parameters

```python
def add(a, b):
    print(a + b)

add(10, 20)
add(5, 7)
```

- Arguments are assigned to parameters according to their positions.

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student("Ram", 20)
```

- return statement : The return statement sends a result back to the code that called the function.

```python
def add(a, b):
    return a + b
```

```python
result = add(10, 20)

print(result)
print(result * 2)
```

- Returning multiple values : Python can return multiple values, typically packaged as a tuple.

```python
def calculate(a, b):
    total = a + b
    difference = a - b
    product = a * b

    return total, difference, product
```

```python
x, y, z = calculate(10, 5)

print(x)
print(y)
print(z)
```

**Note :** Once Python reaches return, the function stops executing.

```python
def check_number(n):
    if n < 0:
        return "Negative number"

    return "Zero or positive"

print(check_number(-5))
print(check_number(10))
```

#### Arguments<br>

- Default Arguments : A default argument provides a value that Python uses when the caller does not supply that argument.

```python
def greet(name="Guest"):
    print("Hello", name)
```

```python

greet()
greet("Ram")
```

- Multiple default arguments

```python
def student(name, age=18, course="Python"):
    print(name, age, course)
```

```python

student("Ram")
student("Sita", 21)
student("Hari", 22, "Data Science")
```

**Important rule:** Parameters without default values must come before parameters with default values.

```python
def example(a=10, b):
    return a + b
```

- Keyword arguments : Keyword arguments let you specify parameter names when calling a function.

```python
def student(name, age, course):
    print(name, age, course)

student(age=20, course="Python", name="Ram")
```

the argument order does not matter when all arguments are provided by keyword.

#### *args — variable positional arguments
Sometimes you do not know how many arguments a function will receive.<br>
*args allows a function to accept any number of additional positional arguments. Inside the function, args is a tuple.

```python
def add_all(*args):
    print(args)
    print(sum(args))
```

```python
add_all(10, 20)
add_all(1, 2, 3, 4, 5)
```

Example : finding the average

```python
def average(*numbers):
    if len(numbers) == 0:
        return 0

    return sum(numbers) / len(numbers)
```

```python
print(average(10, 20, 30))
print(average(5, 15))
print(average())
```

```python
def show_scores(name, *scores):
    print("Student:", name)
    print("Scores:", scores)
```

```python
show_scores("Ram", 80, 90, 85)
```

#### **kwargs — variable keyword arguments
**kwargs lets a function accept any number of additional keyword arguments.

```python
def student_info(**kwargs):
    print(kwargs)
```

```python

student_info(name="Ram", age=20, course="Python")
```

```python
def example(name, *args, **kwargs):
    print("Name:", name)
    print("Extra positional:", args)
    print("Extra keyword:", kwargs)
```

```python
example("Ram", 10, 20, age=20, city="Kathmandu")
```

#### Local and global scope
Scope determines where a variable can be accessed.

- Local variables : A local variable is created inside a function and is generally accessible only within that function.

```python
def my_function():
    message = "Hello"
    print(message)
my_function()
```

Trying to access message outside the function causes a NameError.

```python
def my_function():
    message = "Hello"

my_function()
print(message)
```

- Global variables : A global variable is defined outside functions and can usually be read from within them.

```python
name = "Ram"

def greet():
    print("Hello", name)

greet()
print(name)
```

```python
count = 0

def increase():
    global count
    count += 1

increase()
increase()

print(count)
```

- nonlocal keyword : nonlocal is used in a nested function to modify a variable defined in an enclosing function, but not in the global scope.

```python
def outer():
    count = 0 #<-----non-local variable

    def inner():
        nonlocal count
        count += 1

    inner()
    inner()

    print(count)

outer()
```

#### Lambda functions
A lambda is a small anonymous function written using the lambda keyword.<br>
syntax:<br>
    lambda parameters: expression

```python
square = lambda x: x * x

print(square(5))
print(square(8))
```

```python
add = lambda a, b: a + b

print(add(10, 20))
```

```python
students = [
    ("Ram", 80),
    ("Sita", 95),
    ("Hari", 70)
]

students.sort(key=lambda student: student[1])
#student[1] tells Python to sort by the score, which is the second element in each tuple.
print(students)
```

#### Recursive functions
A recursive function calls itself to solve a smaller version of the same problem.<br>
A recursive function usually needs:
1. Base case: Stops the recursion.
2. Recursive case: Calls the function again with a smaller or simpler input.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```

- Docstrings : A docstring is a string written at the beginning of a function body to explain its purpose.

```python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b
```

```python
print(add(10, 20))
print(add.__doc__)
```

Docstrings help other programmers understand what a function does, what inputs it expects, and what it returns.

```python
def calculate_area(length, width):
    """
    Calculate the area of a rectangle.

    Args:
        length: The length of the rectangle.
        width: The width of the rectangle.

    Returns:
        The area of the rectangle.
    """
    return length * width
```

```python
print(calculate_area.__doc__)
```

- Function annotations : Annotations let you document the expected types of parameters and return values.

```python
def add(a: int, b: int) -> float:
    return a + b

print(add(10, 20))
print(add.__annotations__)
```

- Passing functions as arguments :<br>
In Python, functions are objects. You can pass one function into another function as an argument.

```python
def square(x):
    return x * x

def apply_function(func, value):
    return func(value)

print(apply_function(square, 5))
```
