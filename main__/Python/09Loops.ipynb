### Loops
Loops in Python allow you to execute a block of code repeatedly without writing the same code multiple times.

For example, instead of writing:

```python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
```

we can use a loop:

```python
for i in range(5):
    print("Hello")
```

- for loop :<br>
A for loop is used to iterate over a sequence, such as a list, tuple, string, set, dictionary, or range of numbers.

```python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
```

```python
for letter in "Python":
    print(letter)
```

Example : Calculate the sum of numbers

```python
numbers = [10, 20, 30, 40]
total = 0

for number in numbers:
    total = total + number

print(total)
```

- range ( ) function :<br>
The range ( ) function generates a sequence of integers. It is commonly used when you want a loop to run a specific number of times.

ways to use range ( ) <br>
- A. range(stop)

```python
for i in range(5):
    print(i)
```

- B. range(start, stop)

```python
for i in range(2, 6):
    print(i)
```

- range(start, stop, step)

```python
for i in range(2, 11, 2):
    print(i)
```

- Counting backward : 
we can use a negative step to count backward.

```python
for i in range(5, 0, -1):
    print(i)
```

**Notice  :** that 0 is not printed because the stop value is excluded.<br>
range( ) does not include its stopping value.

- while loop : A while loop repeatedly executes code as long as a condition is True.

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

Example : User input with a while loop

```python
password = ""

while password != "python123":
    password = input("Enter password: ")
    if password != "python123":
        print("Incorrect password Try again.")

print("Access granted")
```

The loop continues until the entered password matches "python123".<br>
This is a basic demonstration; real applications should not store passwords directly in source code.

- Infinite loops : An infinite loop occurs when the condition never becomes False.

```python
i = 1

while i <= 5:
    print(i)
```

|Feature	|for loop	|while loop|
|---|---|---|
|Used when	|Iterating over a sequence or range	|Repeating while a condition is true|
|Number of repetitions	|Often known in advance	|Often depends on a condition|
|Updating |variables	Usually handled by iteration	|Often requires manual updating|
|Risk of infinite loop	|Lower for ordinary finite sequences	|Higher if the condition never changes|

```python
for i in range(3):
    print(i)
```

```python
i = 0

while i < 3:
    print(i)
    i += 1
```

- break statement : The break statement immediately terminates the nearest enclosing loop.

```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

- example: Search for an item

```python
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    if number == 30:
        print("Number found")
        break
```

- continue statement : The continue statement skips the remaining code in the current iteration and moves to the next iteration.

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

When i equals 3, Python skips print(i), but the loop continues with 4 and 5.

example : Print only odd numbers

```python
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
```

- pass statement : The pass statement does nothing. <br>
It acts as a placeholder when Python requires a statement but you do not want to implement the code yet.

```python
for i in range(5):
    pass
```

### Nested loops
A nested loop is a loop inside another loop.<br>
The inner loop completes its iterations for each iteration of the outer loop.

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

Example : Print a star pattern

```python
for i in range(1, 5):
    for j in range(i):
        print("*", end="")
    print()
```

Example : Multiplication table

```python
for i in range(1, 3):
    for j in range(1, 4):
        print(f"{i} x {j} = {i * j}")
    print()
```

- enumerate ( ) function : The enumerate() function allows you to access both the index and the value of each item while looping.

```python
fruits = ["apple", "banana", "mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

without enumerate

```python
fruits = ["apple", "banana", "mango"]

for i in range(len(fruits)):
    print(i, fruits[i])
```

**The First approach is generally cleaner and easier to read.**

example: Number a list of students

```python
students = ["Ram", "Sita", "Hari"]

for roll_no, student in enumerate(students, start=1):
    print(f"{roll_no}. {student}")
```

- zip( ) function : The zip( ) function combines corresponding items from two or more iterables into tuples.

```python
names = ["Ram", "Sita", "Hari"]
ages = [20, 22, 21]

for name, age in zip(names, ages):
    print(name, age)
```

Example : Combine three lists

```python
names = ["Ram", "Sita", "Hari"]
ages = [20, 22, 21]
cities = ["Kathmandu", "Pokhara", "Lalitpur"]

for name, age, city in zip(names, ages, cities):
    print(f"{name}, {age}, {city}")
```

By default, zip( ) stops when the shortest iterable runs out of items.

```python
names = ["Ram", "Sita", "Hari"]
marks = [80, 90]

for name, mark in zip(names, marks):
    print(name, mark)
```
