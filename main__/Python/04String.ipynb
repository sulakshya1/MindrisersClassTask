### String
a string is a sequence of characters used to store text. Strings can contain letters, numbers, spaces, symbols, and special characters.

```python
name = "Python"
message = 'Hello, World!'
```

Both single quotes (' ') and double quotes (" ") can define strings.

#### String indexing
Indexing means accessing a single character in a string.

```python
word = "Python"

print(word[0])
print(word[1])
print(word[5])
```

|Negative index|	-6|	-5|	-4|	-3|	-2|	-1|

```python
word = "Python"

print(word[-1])
print(word[-2]) 
```

#### String slicing
Slicing extracts part of a string.<br>

string [start : stop : step ]
- start: index where extraction begins, included.
- stop: index where extraction ends, excluded.
- step: number of positions to move each time.

```python
word = "Python"

print(word[0:3])   
print(word[2:5])   
print(word[:4])    
print(word[2:])    
print(word[::2])   
print(word[::-1])
print(word[1:5:2])
```

#### Common Python string methods
String methods perform operations on text. Most string methods return a new string or a result without modifying the original string.<br>
- strip( ) — remove surrounding whitespace
- split( ) — divide a string into a list
- join( ) — combine strings
- replace( ) — replace text
- find( ) — find the position of text
- upper( ) — convert to uppercase
- lower( ) — convert to lowercase
- capitalize( ) — Capitalize the first character
- title( ) — Capitalize the first letter of each word
- casefold( ) — Convert to a form suitable for caseless comparison
- swapcase( ) — Reverse uppercase and lowercase

```python
fruits = "apple,banana,mango"
words = ["Python", "is", "easy"]
text = "python programming"
animal = "cat cat cat"

print(text.strip())
print(fruits.split(","))
print("-".join(words))
print(animal.replace("cat", "dog", 2))
print(text.find("Python"))
print(text.find("g"))
print(text.upper())
print(text.lower())
print(text.capitalize())
print(text.title())
print(text.swapcase())
```
