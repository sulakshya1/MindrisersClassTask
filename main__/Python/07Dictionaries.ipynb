### Dictionaries :
A dictionary in Python is a data structure that stores information in key-value pairs. <br>It allows you to store, access, update, and remove data using meaningful keys instead of numeric indexes.

```python
student = {
    "name": "Ram",
    "age": 21,
    "course": "Computer Science",
    "marks": 85
}
type(student)
```

- Creating dictionaries : we can create a dictionary using curly braces { } or the dict( ) function.

```python

student = {
    "name": "Ram",
    "age": 21,
    "marks": 85
}
student2 = dict(name="Sita", age=20, marks=90)
empty_dict = {}

print(student)
print(student2)
print(empty_dict)
```

Important rules:
- Dictionaries store data as key-value pairs.
- Keys must be unique.
- Keys must be hashable, such as strings, numbers, or tuples containing only hashable elements.
- Values can be of any data type, including lists, dictionaries, and sets.
- Dictionaries preserve insertion order in modern Python.
- Dictionaries are mutable, meaning you can change them after creation.

Accessing dictionary values : we can access a value using its key within square brackets [ ]

```python
student = {"name": "Ram", "age": 21, "marks": 85}

print(student["name"])
print(student["age"])
print(student["marks"])
```

However, accessing a key that does not exist raises a KeyError.

```python
print(student["address"])
```

- Using get( ) : The get( ) method retrieves a value without raising an error when the key is missing.

```python
print(student.get("name"))
print(student.get("address"))
print(student.get("address", "Not available"))
```

- Adding and updating values : 
You can add a new key-value pair or update an existing value using square brackets.

```python
student = {"name": "Ram", "age": 21}

# Add a new key
student["course"] = "Python"

# Update an existing key
student["age"] = 22

print(student)
```

- update( ) : 
The update( ) method adds multiple key-value pairs or changes existing ones.

```python
student = {"name": "Ram", "age": 21}
student.update({"age": 22, "marks": 90})

print(student)
```

- Dictionary methods: keys( ), values( ), and items( ) ,<br>These three methods are frequently used when working with dictionaries.

```python
student = {
    "name": "Ram",
    "age": 21,
    "marks": 85
}

print(student.keys())
print(student.values())
print(student.items())
```

|Method	|Purpose|
|---|---|
|keys( )	|Returns all keys|
|values( )|	Returns all values|
|items( )|	Returns key-value pairs as tuples|

- Converting results into lists : 
These methods return dictionary views, not ordinary lists. You can convert them when needed.

```python
print(list(student.keys()))
print(list(student.values()))
print(list(student.items()))
```

- Looping through a dictionary

```python
for key in student.keys():
    print(key)
```

```python
for value in student.values():
    print(value)
```

```python
for key, value in student.items():
    print(key, ":", value)
```

#### Removing dictionary elements

- pop(key) : Removes a specified key and returns its value.

```python
student = {"name": "Ram", "age": 21, "marks": 85}
removed_value = student.pop("age")

print(removed_value)
print(student)
```

- popitem( ) : Removes and returns the last inserted key-value pair.

```python
student = {"name": "Ram", "age": 21, "marks": 85}

print(student.popitem())
print(student)
```

- del : Deletes a specified key-value pair.

```python
student = {"name": "Ram", "age": 21}
del student["age"]

print(student)
```

- clear( ) : Removes all items but keeps the dictionary itself.

```python
student = {"name": "Ram", "age": 21}
student.clear()

print(student)
```

- Checking whether a key exists

```python
student = {"name": "Ram", "age": 21}

print("name" in student)
print("marks" in student)
print("marks" not in student)
```

- sorted( ) : Sorts dictionary keys and returns them in a list.

```python
marks = {"Sita": 90, "Ram": 85, "Hari": 88}

print(sorted(marks))
print(sorted(marks.values()))
```

- len( ) :  Returns the number of key-value pairs.

```python
student = {"name": "Ram", "age": 21, "marks": 85}

print(len(student))
```

### Nested dictionaries
A nested dictionary is a dictionary stored inside another dictionary.

This is useful when storing information about multiple students, employees, products, or other records.

```python
students = {
    "student1": {
        "name": "Ram",
        "age": 21,
        "marks": 85
    },
    "student2": {
        "name": "Sita",
        "age": 20,
        "marks": 92
    }
}
```

```python
print(students["student1"]["name"])
print(students["student2"]["marks"])
```

updating nested values

```python
students["student1"]["marks"] = 90

print(students["student1"]["marks"])
```

Adding a nested record

```python
students["student3"] = {
    "name": "Hari",
    "age": 22,
    "marks": 88
}
```

```python
for key, value in students.items():
    print(key, ":", value)
```

### Dictionary keys and values: important rules

- Duplicate keys<br>
If the same key appears more than once, the last value wins.

```python
data = {"age": 20, "age": 25}

print(data)
```

- Different data types as values <br>
Dictionaries can contain strings, numbers, booleans, lists, and other dictionaries as values.

```python
data = {
    "name": "Ram",
    "age": 21,
    "passed": True,
    "marks": [80, 85, 90],
    "address": {"city": "Kathmandu"}
}
```

- Invalid dictionary keys<br>
Lists and dictionaries cannot be keys because they are mutable and unhashable.

```python
# Invalid
data = {[1, 2]: "value"}
```

```python

# Valid
data = {(1, 2): "value"}
```
