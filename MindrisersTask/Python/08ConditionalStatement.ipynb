### Conditional Statements

Conditional statements allow a program to make decisions based on whether a condition is True or False.

- if statement : 
The if statement executes a block of code only when its condition is True.

```python
age = 20

if age >= 18:
    print("You are an adult.")
```

- if-else statement : 
The else statement provides an alternative when the if condition is False.

```python
age = 16

if age >= 18:
    print("You can vote.")
else:
    print("You cannot vote yet.")
```

- if-elif-else statement

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
elif marks >= 60:
    print("Grade D")
else:
    print("Grade F")
```

- Nested if statements : A nested if is an if statement inside another if statement. It is useful when a decision depends on another decision.

```python
age = 20
registered = True

if age >= 18:
    if registered:
        print("You can vote.")
    else:
        print("Please register to vote.")
else:
    print("You are too young to vote.")
```

#### Comparison operators

Comparison operators compare two values and return either True or False.

|Operator	|Meaning	|Example	|Result|
|---|---|---|---|
|==	|Equal to	|5 == 5	|True|
|!=	|Not equal to	|5 != 3	|True|
|>	|Greater than	|8 > 5	|True|
|<	|Less than	|3 < 7	|True|
|>=	|Greater than or equal to	|5 >= 5	|True|
|<=	|Less than or equal to	|4 <= 3	|False|

```python
password = "python123"

if password == "python123":
    print("Login successful")
else:
    print("Incorrect password")
```

### Logical operators

Logical operators combine or modify conditions.

|Operator|	Meaning	|Example	|Result|
|---|---|---|---|
|and|	Both conditions must be true|	5 > 2 and 8 > 4	|True|
|or	|At least one condition must be true|	5 < 2 or 8 > 4|	True|
|not|	Reverses the truth value	|not (5 > 2)	|False|

- and

```python
age = 22
has_license = True

if age >= 18 and has_license:
    print("You can drive.")
else:
    print("You cannot drive.")
```

- or

```python
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("It is the weekend.")
else:
    print("It is a working day.")
```

- not

```python
is_raining = False

if not is_raining:
    print("You can go outside.")
else:
    print("Take an umbrella.")
```

- Combining comparison and logical operators

```python
age = 25

if age >= 18 and age <= 60:
    print("Eligible")
else:
    print("Not eligible")
```

```python
# login Statement
username = "admin"
password = "12345"

if username == "admin" and password == "12345":
    print("Login successful")
else:
    print("Invalid username or password")
```

- pass statement : The pass statement does nothing. It acts as a placeholder when you want to write a condition but have not decided what code to put inside it yet.

```python
age = 20

if age >= 18:
    pass
else:
    print("You are underage.")
```

- shorthand if-else expression : Python also allows a simple conditional expression on one line.

```python
age = 17
result = "Adult" if age >= 18 else "Minor"

print(result)
```
