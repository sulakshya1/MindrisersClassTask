### Lists
A list in Python is a collection used to store multiple items in a single variable. Lists are ordered, changeable (mutable), and allow duplicate values.<br>
For example, instead of creating separate variables for five student names, you can store them in one list.<br>
it also allow duplicate values.


```python
students = ["Ram", "Sita", "Hari", "Gita", "Ram"]
print(students)
```

#### Creating a list
Use square brackets [ ] to create a list. Separate each item with a comma.

```python
# List of strings
names = ["Ram", "Sita", "Hari"]

# List of integers
numbers = [10, 20, 30, 40]

# List of floating-point numbers
prices = [10.5, 20.75, 30.0]

# Mixed data types
mixed = ["Ram", 20, 85.5, True]

# Empty list
empty_list = []
```

#### Indexing a list
Indexing means accessing an individual item using its position.<br>
Python indexing starts at 0, not 1.

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]
```

```python

print(fruits[0])
print(fruits[2])
print(fruits[-1])
print(fruits[-2])
```

#### Slicing a list
Slicing means extracting a portion of a list.

```python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
print(numbers[:])
print(numbers[::2])
print(numbers[::-1])
```

For numbers[1 : 4 ], Python selects indexes 1, 2, and 3, but not index 4.<br>
For numbers[ : :-1 ], the step is -1, so the list is read backward.

#### Updating list elements
Lists are mutable, which means you can change their elements after creating them.

```python
fruits = ["Apple", "Banana", "Mango"]

fruits[1] = "Orange"

print(fruits)
```

we can also update multiple elements using slicing.

```python
numbers = [10, 20, 30, 40]
numbers[1:3] = [200, 300]

print(numbers)
```

#### Important list methods
Python provides built-in list methods to add, remove, rearrange, and count items.

- append() — add one item,
Adds a single item to the end of the list.

```python
fruits = ["Apple", "Banana"]
fruits.append("Mango")

print(fruits)
```

- extend() — add multiple items,
Adds each item from another iterable to the end of the list.

```python
numbers = [1, 2]
numbers.extend([3, 4, 5])

print(numbers)
```

- insert() — add an item at a specific position

```python
fruits = ["Apple", "Mango"]
fruits.insert(1, "Banana")

print(fruits)
```

- remove() — remove an item by value

```python
numbers = [10, 20, 30, 20, 40]
numbers.remove(20)

print(numbers)
```

- pop() — remove an item by index

```python
numbers = [10, 20, 30, 40]
x = numbers.pop()

print(x)
print(numbers)
```

```python
numbers = [10, 20, 30, 40]
x = numbers.pop(1)

print(x)
print(numbers)
```

- sort() — sort items,
Sorts the original list in ascending order by default.

```python
numbers = [50, 10, 40, 20, 30]
numbers.sort()

print(numbers)
```

for Decending order

```python
numbers.sort(reverse=True)

print(numbers)
```

- reverse() — reverse the list

```python
numbers.reverse()

print(numbers)
```

- count() — count occurrences

```python
numbers = [10, 20, 10, 30, 10, 40]

print(numbers.count(10))
print(numbers.count(20))
print(numbers.count(50))
```

- len() — find the number of items

```python
numbers = [10, 20, 30, 40]

print(len(numbers))
```

- copy() — create a copy

```python
original = [10, 20, 30]

new_list = original.copy()
new_list.append(40)

print(original)
print(new_list)
```
