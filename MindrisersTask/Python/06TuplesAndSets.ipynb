## Tuples and Sets
Tuples and sets are built-in Python data structures used to store multiple values in a single variable. However, they behave differently.<br>
- Tuple: An ordered collection that cannot be changed after creation.
- Set: An unordered collection of unique elements that can be changed after creation.

### Tuples
A tuple is an ordered collection of items. It allows duplicate values and can contain different data types.

```python
# Empty tuple
t1 = ()

# Tuple of integers
t2 = (10, 20, 30, 40)

# Tuple of strings
t3 = ("apple", "banana", "mango")

# Mixed data types
t4 = (10, "Python", 3.14, True)

print(t2)
print(t4)
```

#### Tuple packing
Tuple packing means grouping multiple values into a single tuple.

```python
student = "Ram", 20, "Computer Science"

print(student)
print(type(student))
```

#### Tuple unpacking
Tuple unpacking means assigning the individual elements of a tuple to separate variables.

```python
student = ("Ram", 20, "Computer Science")
name, age, course = student

print(name)
print(age)
print(course)
```

Use * to collect remaining values

```python
numbers = (10, 20, 30, 40, 50)
first, * middle, last = numbers

print(first)
print(middle)
print(last)
```

#### Accessing tuple elements
Tuples use indexing, starting from 0.

```python
fruits = ("apple", "banana", "mango", "orange")

print(fruits[0])
print(fruits[2]) 
print(fruits[-1])
print(fruits[-2])
```

#### Tuple slicing
Slicing extracts a portion of a tuple<br>
Syntax : tuple [ start : stop : step ]

```python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
print(numbers[::2])
print(numbers[::-1])
```

- count() method :
The count() method returns how many times a value appears in a tuple.

```python
numbers = (10, 20, 10, 30, 10, 40)

print(numbers.count(10))
print(numbers.count(20))
print(numbers.count(50))
```

- index() method : 
The index() method returns the index of the first occurrence of a specified value.

```python
fruits = ("apple", "banana", "mango", "banana")

print(fruits.index("banana"))
print(fruits.index("mango"))
#If the value does not exist, Python raises a ValueError.
```

search range

```python
numbers = (10, 20, 30, 20, 40)

print(numbers.index(20, 2))
```

#### Other useful tuple operations
|Operation	|Example	|Result|

### Sets
A set is a collection of unique elements. Sets are useful when you want to remove duplicates or perform mathematical set operations.

```python
numbers = {10, 20, 30, 40}
values = {10, 20, 10, 30, 20}
# Duplicate values are removed
print(values)
type(values)
```

```python
a = set()
b = {}

print(type(a))
print(type(b))
# Use set() to create an empty set. Curly braces {} create an empty dictionary instead.
```

- add() : Adds one element to a set.

```python
fruits = {"apple", "banana"}
fruits.add("mango")
#Adding an element that already exists does not create a duplicate.
fruits.add("banana")

print(fruits)
```

- update() : The update() method adds multiple elements from an iterable, such as a list, tuple, or another set.

```python
numbers = {10, 20}
numbers.update([30, 40, 50])

print(numbers)
```

We can also pass multiple iterables:

```python
numbers = {1, 2}
numbers.update([3, 4], (5, 6), {7, 8})

print(numbers)
```

- remove(element) : Removes the specified element. Raises a KeyError if it does not exist.

```python
numbers = {10, 20, 30}
numbers.remove(20)

print(numbers)
```

- discard(element) : Removes the specified element if present. Does nothing if it does not exist.

```python
numbers = {10, 20, 30}
numbers.discard(50)

print(numbers)
```

- pop() : Removes and returns an arbitrary element. You should not rely on which element it removes.

```python
numbers = {10, 20, 30}
removed = numbers.pop()

print(removed)
print(numbers)
```

- clear() : Removes all elements from the set.

```python
numbers = {10, 20, 30}
numbers.clear()

print(numbers)
```

- Union : union() and | Union combines all elements from two sets, removing duplicates.

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A.union(B))
print(A | B)
```

- Intersection : intersection() and & Intersection returns only elements present in both sets.

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A.intersection(B))
print(A & B)
```

- Difference : difference() and - Difference returns elements in the first set that are not in the second set.

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A.difference(B))
print(B.difference(A))
```

- Symmetric difference : symmetric_difference() and ^ Returns elements that belong to either set, but not both.

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A.symmetric_difference(B))
print(A ^ B)
```

- issubset() : Checks whether every element of one set is present in another set.

```python
A = {1, 2}
B = {1, 2, 3, 4}

print(A.issubset(B))
print(A <= B)  
```

- issuperset() : Checks whether a set contains all elements of another set.

```python
A = {1, 2, 3, 4}
B = {1, 2}

print(A.issuperset(B))
print(A >= B)   
```

- isdisjoint() : Returns True if two sets have no common elements.

```python
A = {1, 2, 3}
B = {4, 5, 6}
C = {3, 4, 5}

print(A.isdisjoint(B))
print(A.isdisjoint(C))
```

- copy() : Creates a shallow copy of a set.

```python
A = {10, 20, 30}
B = A.copy()

B.add(40)

print(A)
print(B)
```

- len() and membership testing

```python
numbers = {10, 20, 30, 40}

print(len(numbers))
print(20 in numbers)
print(50 not in numbers)
```

**_NOTE:_** Sets do not support indexing or slicing.

|Mutable|	No	|Yes|
|Indexing|	Yes	|No|
|Slicing|	Yes	|No|
|Empty creation| ( )	|set( )|
|Main use|	Fixed collections of values|	Unique values and set operations|
