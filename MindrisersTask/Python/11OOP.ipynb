### Object-Oriented Programming (OOP)

Object-Oriented Programming (OOP) is a programming approach in which we organize code using classes and objects. It helps us write reusable, organized, and maintainable code.

#### Classes and Objects
**What is a class?**<br >
&ensp;&ensp;A class is a blueprint or template for creating objects. It defines the data and behaviors that its objects can have.<br>
<br>
**What is an object?**<br>
&ensp;&ensp;An object is an instance of a class. Each object can have its own data.

```python
class Student:
    name = "Ram"
    age = 20

# Creating an object
s1 = Student()

print(s1.name)
print(s1.age)
```

we can create multiple objects from the same class.

```python
class Student:
    pass

s1 = Student()
s2 = Student()

print(type(s1))
print(type(s2))
```

#### The __init__() Method
The __init__() method is used to initialize an object's attributes when the object is created.<br>
It is commonly called the constructor in beginner explanations, although technically it initializes an object after it has been created.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Ram", 20)
s2 = Student("Sita", 22)

print(s1.name, s1.age)
print(s2.name, s2.age)
```

- self Parameter :<br>
self refers to the current instance of a class. It allows a method to access or modify that object's attributes and call its other instance methods.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, my name is", self.name)

s1 = Student("Ram")
s2 = Student("Sita")
```

```python
s1.greet()
s2.greet()
```

#### Methods
A method is a function defined inside a class.<br>
Python OOP commonly uses three kinds of methods.

- Instance methods : Instance methods work with individual objects. They normally use self.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(self.name, self.marks)

s1 = Student("Ram", 85)
s1.display()
```

- Class methods : Class methods operate on the class itself and receive cls as their first parameter. They are defined using @classmethod.<br>
&ensp;&ensp;Class methods are useful for working with class-wide data or creating alternative ways to construct objects.

```python
class Student:
    school = "ABC College"

    @classmethod
    def show_school(cls):
        print(cls.school)

Student.show_school()
```

- Static methods : Static methods do not automatically receive self or cls. They are defined using @staticmethod.

```python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
```

#### Instance Variables and Class Variables
Variables inside a class can belong to individual objects or be shared at the class level.
- Instance variables :<br>
&ensp;&ensp;Instance variables belong to individual objects. They are commonly created using self.
- Class variables :<br>
&ensp;&ensp;Class variables are defined in the class body and are shared through the class unless an instance attribute shadows them.

```python
class Student:
    school = "ABC College"  # Class variable

    def __init__(self, name):
        self.name = name    # Instance variable

s1 = Student("Ram")
s2 = Student("Sita")

print(s1.name)
print(s2.name)

print(s1.school)
print(s2.school)
```

### Inheritance
Inheritance allows one class to acquire attributes and methods from another class.
- Parent class: The class whose features are inherited.
- Child class: The class that inherits those features.
Inheritance promotes code reuse

```python
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

d = Dog()
# The Dog class inherits the eat() method from Animal and defines its own bark() method.
d.eat()
d.bark()
```

#### Types of inheritance

- Single inheritance : 
One child inherits from one parent.

```python
class A:
    pass

class B(A):
    pass
```

- Multiple inheritance : 
One child inherits from multiple parents.

```python
class A:
    pass

class B:
    pass

class C(A, B):
    pass
```

- Multilevel inheritance : 
A child class becomes the parent of another class.

```python
class A:
    pass

class B(A):
    pass

class C(B):
    pass
```

- Hierarchical inheritance : Multiple child classes inherit from the same parent.

```python
class Animal:
    pass

class Dog(Animal):
    pass

class Cat(Animal):
    pass
```

- Hybrid inheritance : 
A combination of two or more inheritance patterns.

 #### super( ) Function
The super( ) function allows a class to access methods from the next class in the method resolution order (MRO). <br>It is commonly used to call a parent's implementation.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Animal:", self.name)

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

    def show(self):
        super().display()
        print("Breed:", self.breed)

d = Dog("Tommy", "Labrador")
d.show()
```

#### Method Overriding
Method overriding occurs when a child class defines a method with the same name as a method inherited from its parent.<br>
The child implementation is used when that method is called on a child object.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        super().sound() # whith super() we can call the parent class method from the child class
        print("Dog barks")

a = Animal()
d = Dog()

d.sound()
```

### Polymorphism
Polymorphism means “many forms.” In programming, it allows the same operation or method call to behave differently for different objects.

Example : Polymorphism through overriding

```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
```

Example : Duck typing<br>
- duck typing : an object can be used in a particular context if it provides the required behavior, even if it does not inherit from a specific class.

```python
class Car:
    def move(self):
        print("Car is moving")

class Person:
    def move(self):
        print("Person is walking")

def start_moving(obj):
    obj.move()

start_moving(Car())
start_moving(Person())
```

Example : Operator overloading <br>
`+` behaves differently depending on operand type


```python
print(10 + 20)
print("Hello " + "World")
print([1, 2] + [3, 4])
```

### Encapsulation
Encapsulation means combining data and methods in a class and controlling how that data is accessed or modified.

- Public attributes : 
Public attributes can be accessed from outside the class.

```python
class Student:
    def __init__(self, name):
        self.name = name

s = Student("Ram")
print(s.name)
```

- Protected-by-convention attributes : 
A single leading underscore indicates that an attribute is intended for internal use.

```python
class Student:
    def __init__(self):
        self._marks = 80
```

- Name-mangled attributes : 
A double leading underscore triggers name mangling

```python
class Student:
    def __init__(self, marks):
        self.__marks = marks

    def display(self):
        print(self.__marks)

s = Student(85)
s.display()
```

- Using properties to control access :
A common Python approach is to use @property to validate data when it is assigned.

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):
        if 0 <= value <= 100:
            self._marks = value
        else:
            raise ValueError("Marks must be between 0 and 100")

s = Student(85)
print(s.marks)

s.marks = 90
print(s.marks)
```

#### Method Resolution Order (MRO)
MRO determines the order in which Python searches classes for a method or attribute, particularly when inheritance involves multiple parents.

```python
class A:
    def show(self):
        print("A")

class B(A):
    pass

class C(B):
    pass

print(C.mro())
```

### Example Combining OOP Concepts

```python
class Employee:
    company = "ABC Technologies"

    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

    def work(self):
        print(self.name, "is working")

    def get_salary(self):
        return self.__salary


class Developer(Employee):
    def __init__(self, name, salary, language):
        super().__init__(name, salary)
        self.language = language

    def work(self):
        print(self.name, "is coding in", self.language)


class Analyst(Employee):
    def work(self):
        print(self.name, "is analyzing data")


employees = [
    Developer("Ram", 50000, "Python"),
    Analyst("Sita", 45000)
]

for employee in employees:
    employee.work()
    print("Salary:", employee.get_salary())
    print("Company:", employee.company)
```

### Error Handling
Error handling is the process of detecting and managing errors so that a program can respond appropriately instead of stopping unexpectedly.

Types of errors in Python
- Syntax errors : These occur when Python code violates the language's syntax rules.
- Logical errors : These occur when the code runs without an exception but produces the wrong result.
- Runtime errors (exceptions) : These occur while the program is executing.

|Exception	|Cause|
|---|---|
|ZeroDivisionError|	Dividing by zero|
|ValueError	|Correct type of input operation, but an invalid value|
|TypeError	|An operation receives an inappropriate type|
|NameError	|Using an undefined variable name|
|IndexError	|Accessing a list index that does not exist|
|KeyError	|Accessing a missing dictionary key|
|FileNotFoundError|	Trying to open a file that does not exist|
|AttributeError	|Accessing an attribute an object does not have|
|ImportError	|An import cannot be completed|

#### Handeling
try and except blocks :Python uses try and except to handle exceptions.
- try: Contains code that might raise an exception.
- except: Handles a matching exception if it occurs.

```python
try:
    result = 100 / int(input("Enter divisor: "))

except (ValueError, ZeroDivisionError):
    print("Please enter a nonzero integer.")

else:
    print("Result:", result)
```

- The finally block : The finally block runs whether an exception occurs or not, including when an exception is handled by an except block.

```python
try:
    number = int("abc")

except ValueError:
    print("Invalid conversion.")

finally:
    print("Program finished.")
```

general structure

```python
try:
    # Code that might raise an exception
    result = 10 / 2

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result:", result)

finally:
    print("Execution completed.")
```

- Raising exceptions with raise : The raise statement lets you deliberately generate an exception when a condition is invalid.<br>
For example, suppose a function should accept only positive numbers.

```python
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")

    print("Age:", age)

check_age(10)
```

```python
check_age(-10)
```

- Custom exceptions :
You can define your own exception types by inheriting from Python's Exception class.

```python
class InsufficientBalanceError(Exception):
    pass


def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError(
            "Insufficient balance."
        )

    return balance - amount


try:
    remaining = withdraw(1000, 1500)
    print("Remaining balance:", remaining)

except InsufficientBalanceError as e:
    print(e)
```
