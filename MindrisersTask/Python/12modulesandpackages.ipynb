### Python Modules and Packages

What are modules and packages?<br>
&ensp;&ensp; - A module is a Python file containing reusable code, such as functions, variables, and classes.<br>
&ensp;&ensp; - A package is a way to organize related Python modules into a directory structure.

Module:<br>
&ensp;Example: calculator.py<br>
&ensp;&ensp;Contains functions for addition, subtraction, and other calculations.

Package:<br>
&ensp;Example: mypackage/<br>
&ensp;&ensp;Contains multiple related modules, often with an __init__.py file.

#### Importing modules

Python's import statement lets you use code written in another module.<br>
**Suppose we have a module named calculator.py:**

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
```

```python
import calculator
print(calculator.add(10, 5))
print(calculator.subtract(10, 5))
```

- Using from ... import<br>
You can import specific functions or variables directly.

```python
from calculator import add

print(add(10, 5))
```

|Syntax	|How to use the function|
|---|---|
|import calculator	|calculator.add(2, 3)|
|from calculator import add	|add(2, 3)|
|from calculator import add, subtract	|add(2, 3) or subtract(5, 2)|
|import calculator as c	|c.add(2, 3)|

- Import aliases using as<br>
An alias gives a module or imported object a shorter or alternative name.

```python
import math as m

print(m.sqrt(25))
print(m.pi)
```

Aliases are useful when a module has a long name or when you want a convenient name.

```python
import pandas as pd
import numpy as np
```

- Creating your own custom module<br>
You can create your own modules by saving Python code in a .py file.

```python
import calculator
print(calculator.add(10, 5))
print(calculator.subtract(10, 5))
```

What is a Python package?<br>
&ensp;A package organizes related modules into a directory.

```python
from package.greetings import greet, goodbye
from package.student import get_student

print(greet("Ram"))
print(goodbye("Ram"))
print(get_student())
```

- `__init__`.py : <br>
The` __init__`.py file is associated with Python packages.<br>
It can be used to initialize a package or expose selected functions and classes for easier importing.

```python
from package import greetings

print(greetings.greet("Ram"))
```

#### math module
The math module provides mathematical functions and constants. It is part of Python's standard library, so you do not need to install it with pip.

```python
import math
```

|Function	|Purpose	|Example result|
|---|---|---|
|math.sqrt(25)	|Square root	|5.0|
|math.ceil(4.2)	|Round upward	|5|
|math.floor(4.8)	|Round downward	|4|
|math.factorial(5)	|Factorial	|120|
|math.pow(2, 3)	|Power	|8.0|
|math.fabs(-10)	|Absolute value as a float	|10.0|
|math.gcd(12, 18)|	Greatest common divisor	|6|
|math.trunc(4.9)	|Remove fractional part	|4|

```python
number = 16

print(math.sqrt(number))
print(math.factorial(5))
print(math.ceil(4.1))
print(math.floor(4.9))
```

```python
print(math.pi)
print(math.e)
print(math.sin(math.radians(90)))
print(math.cos(math.radians(0)))
```

#### random module
The random module generates pseudorandom values. It is useful for games, simulations, sampling, and practice exercises.

```python
import random
```

|Function	|Purpose|
|---|---|
|random.random()	|Float from 0.0 up to, but not including, 1.0|
|random.randint(a, b)|	Random integer including both endpoints|
|random.randrange(a, b)|	Random integer from a up to, but not including, b|
|random.choice(sequence)|	Select one item|
|random.shuffle(list)	|Shuffle a list in place|
|random.sample(sequence, k)|	Select k unique items|
|random.uniform(a, b)	|Random float between the bounds|

```python
print(random.randint(1, 10))
```

```python
fruits = ["apple", "banana", "mango", "orange"]

print(random.choice(fruits))
```

```python
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)

print(numbers)
```

#### datetime module
The datetime module handles dates, times, and time differences.

```python
import datetime
```

```python
from datetime import datetime

now = datetime.now()
print(now)
```

```python

from datetime import date
today = date.today()

print(today)
```

```python
birthday = date(1919, 5, 15)

print(birthday)
print(birthday.year)
print(birthday.month)
print(birthday.day)
```

- Format dates using strftime ( )

```python
now = datetime.now()

print(now.strftime("%d-%m-%Y"))
print(now.strftime("%H:%M:%S"))
print(now.strftime("%A"))
```

**Common format codes:**

|Code	|Meaning	|Example|
|---|---|---|
|%d	|Day of month	|27|
|%m	|Month number	|10|
|%Y	|Four-digit year|	2026|
|%y	|Two-digit year	|26|
|%H	|Hour in 24-hour format	|20|
|%I	|Hour in 12-hour format	|08|
|%M	|Minute	|30|
|%S	|Second	|15|
|%A	|Full weekday name	|Tuesday|
|%B	|Full month name	|October|

- Calculating date differences using timedelta

```python
from datetime import date, timedelta

today = date.today()
tomorrow = today + timedelta(days=1)
last_week = today - timedelta(days=7)

print(tomorrow)
print(last_week)
```

#### What is pip?
pip is Python's package installer. It helps you install and manage external libraries that are not included in the Python standard library.<br>
For example, math and datetime come with Python, but libraries such as pandas and requests are typically installed separately.

**Common pip commands**

|Command|	Purpose|
|---|---|
|python -m pip install pandas	|Install a package|
|python -m pip uninstall pandas	|Uninstall a package|
|python -m pip show pandas	|Show package information|
|python -m pip list	|List installed packages|
|python -m pip freeze	|List installed packages and versions|
|python -m pip install --upgrade pandas	|Upgrade a package|
|python -m pip --version	|Display the pip version|
