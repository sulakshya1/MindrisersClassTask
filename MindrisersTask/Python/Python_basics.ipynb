### Indentation in Python

<b>Indentation means adding spaces at the beginning of a line of code.</b><br>
Python uses indentation to identify which statements belong together in a block. Unlike languages such as C or Java,<br>
Python does not use curly braces {} to define these blocks.

```python
if 10 > 5:
    print("10 is greater than 5")
    print("This is inside the if block")
print("This is outside the if block")
```

### Comments in Python
<b>Comments are notes written in code for humans to read. Python ignores comments when executing the program.</b><br>
Python does not have a dedicated multiline comment symbol. You can use # on each line<br>
<b>Triple-quoted strings can span multiple lines, but they are strings, not true comments.<br>
They are often used for documentation, called</b> *docstrings*.

```python
# This is a comment
"""
This is a multiline string.
It can be on multiple lines.
TO create a multiline string Comment, you can use triple quotes (either single or double).
"""
print("Hello")
```

### The print() Function

#### Using sep and end

<b>sep</b> specifies the separator between values.<br>
<b>end</b> specifies what is printed at the end of the statement.

```python
print("apple", "banana", "mango", sep="-")
print("Hello", end=" ")
print("World")
```

### The input() Function
The input() function accepts information entered by the user through the keyboard.<br>
*Even if the user enters a number, Python initially treats the input as a string.*

```python
name = input("Enter your name: ")
print("Hello", name)
```

```python
age = input("Enter your age: ")
print(type(age))
```

To use the input as a number, convert it to integer from 'Str'

```python
age = int(input("Enter your age: "))
print(age + 1)
```

### Variables in Python
Variables are containers for storing data values.Python has no command for declaring a variable.<br>
A variable is created the moment you first assign a value to it.

```python
name = "Ram"
age = 20
height = 5.8
is_student = True
```

#### Assigning and updating variables

```python
x = 10
print(x)

x = 25
print(x)
```

#### Multiple assignments

```python
a, b, c = 10, 20, 30

print(a)
print(b)
print(c)
```

You can also assign the same value to multiple variables:

```python
a = b = c = 100
```

Python is dynamically typed
You do not need to declare a variable's type explicitly. Python determines the type from the assigned value.

```python
x = 10
print(type(x))

x = "Python"
print(type(x))
```

### Naming Rules for Variables

### Excape Character
An escape character is a special character sequence that begins with a backslash ( \ ) and represents a character or formatting instruction that may be difficult to write directly in a string.

|Escape |character|	Meaning	Example|
|-------|----------|---------------|
|\n	|New line	|"Hello\nWorld"|
|\t	|Horizontal tab	|"Hello\tWorld"|
|\\	|Backslash	|"C:\\Python"|
|\'	|Single quotation mark	|'It\'s Python'|
|\"	|Double quotation mark	|"He said \"Hi\""|
|\r |Carriage return	|"Hello\rWorld"|
|\b	|Backspace	|"Hello\bWorld"|
|\0	|Null character	|"A\0B"|

```python
print("Hello\nWorld")
```
