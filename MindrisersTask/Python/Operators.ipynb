### Operators
An operator is a symbol or keyword that performs an operation on values or variables.

#### Types of Python operators
Python operators are commonly divided into seven categories.
| Category | Purpose | Operators |
|---|---|---|
| Arithmetic | Perform mathematical calculations | +, -, *, /, //, %, ** |
| Assignment | Assign or update values | =, +=, -=, *=, /=, etc. |
| Comparison | Compare two values | ==, !=, >, <, >=, <= |
| Logical | Combine or reverse conditions | and, or, not |
| Identity | Check whether objects are the same object | is, is not |
| Membership | Check whether a value exists in a collection | in, not in |
| Bitwise | Operate on individual bits of integers | &, |, ^, ~, <<, >> |

```python
a = 10
b = 5

print(a + b)
print(a > b)
```

#### Arithmetic operators
Arithmetic operators perform mathematical calculations.
|Operator	|Name	|Example	|Result|
|---|---|---|---|
|+	|Addition	|a + b	|13|
|-	|Subtraction	|a - b	|7|
|*	|Multiplication	|a * b	|30|
|/	|Division	|a / b	|3.333...|
|//	|Floor division	|a // b	|3|
|%	|Modulus	|a % b	|1|
|**	|Exponentiation	|a ** b	|1000|

The + operator can also join strings.<br>
" * " can also repeat a string.

```python
print("Hello " + "World")
print("Hi " * 3)
```

#### Assignment operators
Assignment operators assign values to variables or update existing values.
|Operator	|Equivalent expression	|Result|
|---|---|---|
x += 5	|x = x + 5	|15
x -= 3	|x = x - 3	|7
x *= 2	|x = x * 2	|20
x /= 2	|x = x / 2	|5.0
x //= 3	|x = x // 3	|3
x %= 3	|x = x % 3	|1
x **= 2	|x = x ** 2	|100

```python
score = 50
score += 10
score -= 5
print(score)
```

#### Comparison operators
Comparison operators compare two values and return a Boolean value: True or False.

|Operator	|Meaning	|Example	|Result|
|---|---|---|---|
==	|Equal to	|a == b	|False
!=	|Not equal to	|a != b	|True
|>	|Greater than	|a > b	|True
<	|Less than	|a < b	|False
|>=	|Greater than or equal to	|a >= 10	|True
<=	|Less than or equal to	|b <= 5	|True

```python
age = 20

print(age >= 18)
print(age == 20)
print(age != 20)
```

#### Logical operators
Logical operators combine conditions or reverse their truth values.<br>
Python has three logical operators: and, or, and not.

```python
age = 22

print(age >= 18 and age <= 60)
```

##### Truth table
**and** Returns True when both conditions are true.<br>
**or** Returns True when at least one condition is true.<br>
**not** Reverses the truth value of a condition.
|A	|B	|A and B|	A or B	|not A|
|---|---|---|---|---|
|True	|True	|True	|True	|False
|True	|False	|False	|True	|False
|False	|True	|False	|True	|True
|False	|False	|False	|False	|True

```python
day = "Saturday"
print(day == "Saturday" or day == "Sunday")

is_raining = False
print(not is_raining)
```

#### Identity operators
Identity operators check whether two references point to the same object, not merely whether their values are equal.<br>
Python has two identity operators:
- is
- is not

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)
print(a is c)
print(a == c)
```

#### Membership operators
Membership operators check whether a value exists in a collection or sequence.<br>
Python has two membership operators:
- in : Returns True when the value is present.
- not in : Returns True when the value is absent.

```python
fruits = ["apple", "banana", "mango"]

print("banana" in fruits)
print("grapes" in fruits)
```

membership in dictionaries

```python
student = {"name": "Ram", "age": 20}

print("name" in student)
print("Ram" in student)
print("Ram" in student.values())
```

#### Bitwise operators
Bitwise operators perform operations on the individual bits of integers.<br>
For beginners, it is useful to understand the basics, even if you do not use these operators every day.

```python
a = 5 #binaryfor 0101
b = 3 #binaryfor 0011
print(a & b)  # bitwise AND 
```

#### Operator precedence
When an expression contains multiple operators, Python follows rules to determine the order of evaluation.

```python
result = 2 + 3 * 4
print(result)
```

Multiplication occurs before addition, so the calculation is \( 2 + ( 3 * 4) = 14 \).<br>

Common precedence order
From higher to lower precedence:

| Priority | Operator or category | Example |
|---|---|---|
| 1 | Parentheses | (2 + 3) |
| 2 | Exponentiation | 2 ** 3 |
| 3 | Unary operators | -x, ~x |
| 4 | Multiplication, division, floor division, modulus | *, /, //, % |
| 5 | Addition and subtraction | +, - |
| 6 | Bit shifts | <<, >> |
| 7 | Bitwise AND | & |
| 8 | Bitwise XOR | ^ |
| 9 | Bitwise OR | \| |
| 10 | Comparisons, membership, identity | <, in, is |
| 11 | Logical NOT | not |
| 12 | Logical AND | and |
| 13 | Logical OR | or |
